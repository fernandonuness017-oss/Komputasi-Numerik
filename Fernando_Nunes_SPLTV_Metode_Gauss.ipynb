```python

# ==========================================
# PROGRAM KOMPUTASI NUMERIK
# SISTEM PERSAMAAN LINEAR TIGA VARIABEL
# METODE ELIMINASI GAUSS
# ==========================================

# Nama       : Fernando Nunes
# NIM        : 202511420011
# Prodi      : Teknik Informatika
# Mata Kuliah: Komputasi Numerik

print("PROGRAM PENYELESAIAN SPLTV")
print("METODE ELIMINASI GAUSS")
print("Nama: Fernando Nunes")
print("NIM: 202511420011")
print("================================")
```

```python
# Matriks koefisien x, y, dan z
A = [
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
]

# Hasil dari setiap persamaan
B = [10, 11, 12]

# Menampilkan data
print("Matriks A:")
for baris in A:
    print(baris)

print("Matriks B:")
print(B)
```

```python
# Membuat salinan matriks
M = [baris[:] for baris in A]
H = B[:]

# Menentukan jumlah variabel
n = len(H)

# Melakukan eliminasi
for i in range(n):

    # Menukar baris jika pivot nol
    if M[i][i] == 0:
        for k in range(i + 1, n):
            if M[k][i] != 0:
                M[i], M[k] = M[k], M[i]
                H[i], H[k] = H[k], H[i]
                break

    if M[i][i] == 0:
        raise ValueError("Sistem tidak memiliki solusi tunggal.")

    # Menghilangkan elemen di bawah pivot
    for j in range(i + 1, n):
        faktor = M[j][i] / M[i][i]

        for k in range(i, n):
            M[j][k] -= faktor * M[i][k]

        H[j] -= faktor * H[i]

print("Matriks setelah eliminasi:")
for baris in M:
    print(baris)
```

```python
# Menyiapkan tempat hasil variabel
hasil = [0.0] * n

# Menghitung variabel dari bawah ke atas
for i in range(n - 1, -1, -1):

    jumlah = 0

    for j in range(i + 1, n):
        jumlah += M[i][j] * hasil[j]

    hasil[i] = (H[i] - jumlah) / M[i][i]
```

```python
# Menampilkan hasil akhir
print("================================")
print("HASIL PENYELESAIAN SPLTV")
print("================================")

print("Nilai X =", hasil[0])
print("Nilai Y =", hasil[1])
print("Nilai Z =", hasil[2])

print("================================")
print("PROGRAM SELESAI")
```
